```python
# Function to generate a summary report for a DataFrame
def generate_summary_report(df):
    report = []

    # Data Overview
    num_rows, num_cols = df.shape
    report.append(f"Data Overview:\n- Number of Rows: {num_rows}\n- Number of Columns: {num_cols}\n")

    # Column Information
    report.append("Column Information:")
    for col, dtype in zip(df.columns, df.dtypes):
        report.append(f"- {col}: {dtype}\n")

    # Summary Statistics (for numeric columns)
    report.append("Summary Statistics:")
    numeric_cols = df.select_dtypes(include=['number']).columns
    for col in numeric_cols:
        stats = df[col].describe().to_string()
        report.append(f"Summary Statistics for {col} Column:\n{stats}\n")

    # Data Quality
    missing_values = df.isnull().sum()
    report.append("Data Quality (Missing Values):")

    # Create a histogram for missing rates
    missing_rates = (missing_values / num_rows) * 100
    plt.figure(figsize=(12, 6))
    plt.bar(missing_rates.index, missing_rates.values)
    plt.title('Missing Rate by Column')
    plt.xlabel('Column Name')
    plt.ylabel('Missing Rate (%)')
    plt.xticks(rotation=90)
    plt.tight_layout()
    plt.show()
    
    report.append("Data Quality:")
    for col, missing_count in missing_values.items():
        if missing_count > 0:
            report.append(f"- {col}: {missing_count / num_rows} missing rates")

    return '\n'.join(report)

# Generate the summary report
summary_report = generate_summary_report(df)

# Print or save the summary report as needed
print(summary_report)
```
